```python
import pandas as pd
import numpy as np

import requests

# --- Professor Names ---
professor_kauffman = "Christopher Kauffman"
professor_cliff = "Cliff Bakalian"
professor_ting = "Ting Jiang"
professor_fardina = "Fardina Alam"
professor_max = "Maksym Morawski"
# extra
professor_rumeyza = "Rumeyza Karan"
professor_jonathanfernandes = "Jonathan Fernandes"
professor_justin = "Justin Wyss-Gallifent"

# --- API Calls ---
# looked up how to do this (https://stackoverflow.com/questions/61633719/can-you-pull-an-api-call-straight-into-a-pandas-dataframe)
professor_kauffman_reviews = requests.get("https://planetterp.com/api/v1/professor",
  params = {
      "name": professor_kauffman,
      # "courses":
      "reviews": "true"
  })

professor_cliff_reviews = requests.get("https://planetterp.com/api/v1/professor",
  params = {
      "name": professor_cliff,
      # "courses":
      "reviews": "true"
  })

professor_ting_reviews = requests.get("https://planetterp.com/api/v1/professor",
  params = {
      "name": professor_ting,
      # "courses":
      "reviews": "true"
  })

professor_fardina_reviews = requests.get("https://planetterp.com/api/v1/professor",
  params = {
      "name": professor_fardina,
      # "courses":
      "reviews": "true"
  })

professor_max_reviews = requests.get("https://planetterp.com/api/v1/professor",
  params = {
      "name": professor_max,
      # "courses":
      "reviews": "true"
  })

professor_rumeyza_reviews = requests.get("https://planetterp.com/api/v1/professor",
  params = {
      "name": professor_rumeyza,
      # "courses":
      "reviews": "true"
  })

professor_jonathanfernandes_reviews = requests.get("https://planetterp.com/api/v1/professor",
  params = {
      "name": professor_jonathanfernandes,
      # "courses":
      "reviews": "true"
  })

professor_justin_reviews = requests.get("https://planetterp.com/api/v1/professor",
  params = {
      "name": professor_justin,
      # "courses":
      "reviews": "true"
  })

# --- JSONs to Dictionary ---
# to see what's being pulled from API
# looked this up (https://stackoverflow.com/questions/18810777/how-do-i-read-a-response-from-python-requests)
professor_kauffman_reviews_dict = professor_kauffman_reviews.json()
professor_cliff_reviews_dict = professor_cliff_reviews.json()
professor_ting_reviews_dict = professor_ting_reviews.json()
professor_fardina_reviews_dict = professor_fardina_reviews.json()
professor_max_reviews_dict = professor_max_reviews.json()
professor_rumeyza_reviews_dict = professor_rumeyza_reviews.json()
professor_jonathanfernandes_reviews_dict = professor_jonathanfernandes_reviews.json()
professor_justin_reviews_dict = professor_justin_reviews.json()

# --- Professor Courses ---
professor_kauffman_courses = "CMSC216, CMSC330"
professor_cliff_courses = "CMSC250, CMSC330, CMSC389E, HDCC106"
professor_ting_courses = "CMSC351, CMSC417"
professor_fardina_courses = "DATA100, DATA200, CMSC320, DATA602"
professor_max_courses = "CMSC106, CMSC250, CMSC320, CMSC351, CMSC421, DATA602"
professor_rumeyza_courses = "EDHD412"
professor_jonathanfernandes_courses = "DATA100, DATA400, MATH140, MATH141, MATH461, MSML601, STAT100, STAT400, STAT401, STAT410, STAT420, STAT426, STAT498J, SURV410"
professor_justin_courses = "CMSC250, CMSC351, CMSC420, CMSC456, MATH115, MATH130, MATH131, MATH140, MATH140H, MATH141H, MATH206, MATH220, MATH221, MATH240, MATH241, MATH246, MATH401, MATH403, MATH406, MATH410, MATH411, MATH431, MATH456, MATH475"

# --- Clean Dictionary ---
professor_reviews = []

for i in professor_kauffman_reviews_dict["reviews"]:
    professor_reviews.append({
        "professor": professor_kauffman,
        "course": professor_kauffman_courses,
        "review": i.get("review"),
        "rating": i.get("rating")
    })

for i in professor_cliff_reviews_dict["reviews"]:
    professor_reviews.append({
        "professor": professor_cliff,
        "course": professor_cliff_courses,
        "review": i.get("review"),
        "rating": i.get("rating")
    })

for i in professor_ting_reviews_dict["reviews"]:
    professor_reviews.append({
        "professor": professor_ting,
        "course": professor_ting_courses,
        "review": i.get("review"),
        "rating": i.get("rating")
    })

for i in professor_fardina_reviews_dict["reviews"]:
    professor_reviews.append({
        "professor": professor_fardina,
        "course": professor_fardina_courses,
        "review": i.get("review"),
        "rating": i.get("rating")
    })

for i in professor_max_reviews_dict["reviews"]:
    professor_reviews.append({
        "professor": professor_max,
        "course": professor_max_courses,
        "review": i.get("review"),
        "rating": i.get("rating")
    })

for i in professor_rumeyza_reviews_dict["reviews"]:
    professor_reviews.append({
        "professor": professor_rumeyza,
        "course": professor_rumeyza_courses,
        "review": i.get("review"),
        "rating": i.get("rating")
    })

for i in professor_jonathanfernandes_reviews_dict["reviews"]:
    professor_reviews.append({
        "professor": professor_jonathanfernandes,
        "course": professor_jonathanfernandes_courses,
        "review": i.get("review"),
        "rating": i.get("rating")
    })

for i in professor_justin_reviews_dict["reviews"]:
    professor_reviews.append({
        "professor": professor_justin,
        "course": professor_justin_courses,
        "review": i.get("review"),
        "rating": i.get("rating")
    })

df_of_reviews = pd.DataFrame(professor_reviews) # convert to data frame
# df_of_reviews.head(1000)

# --- Data Cleaning ---
df_of_reviews = df_of_reviews.dropna()
df_of_reviews = df_of_reviews.drop_duplicates()

# --- Visualize Reviews By Stars ---
import matplotlib.pyplot as plt

df_of_reviews["rating"].value_counts().sort_index().plot(kind = "bar", color = ["#a68aea", "#6F52BF"])
plt.title("Number of Star Reviews")
plt.xlabel("Stars")
plt.xticks(rotation = 0)
plt.ylabel("Number of Reviews")
plt.show()

# --- Transformer Model ---
# looked up how to do this (https://stackoverflow.com/questions/65499783/apply-transformer-model-to-each-row-in-a-pandas-column)
from transformers import pipeline
classifier = pipeline(model = "nlptown/bert-base-multilingual-uncased-sentiment")

transformer_ratings = []
for i in df_of_reviews["review"]:
  prediction = classifier(i[:512]) # "512" from piazza post @186
  label = prediction[0]["label"]
  rating_prediction = int(label[0])
  transformer_ratings.append(rating_prediction)

df_of_reviews["rating_prediction"] = transformer_ratings
# df_of_reviews.head(1000)

# df_of_reviews = pd.read_csv("professor_reviews.csv")

# --- Plot For Accuracy ---
import seaborn as sns

actual_vs_prediction = pd.crosstab(
    df_of_reviews["rating"],
    df_of_reviews["rating_prediction"]
)

sns.heatmap(actual_vs_prediction, annot = True, fmt = "d", cmap = "Purples")
plt.title("Confusion Matrix")
plt.xlabel("Predicted Rating")
plt.ylabel("Actual Rating")
plt.show()

# --- Average Error ---
# looked up (https://stackoverflow.com/questions/72045199/calculate-mean-absolute-error-for-each-row-of-a-pandas-dataframe)
difference_in_rating = abs(
    df_of_reviews["rating"] - df_of_reviews["rating_prediction"]
)
average_error = difference_in_rating.mean()
average_error = average_error.round(2)
print("Average Error: ", average_error)
print("On average, transformer was off by", average_error, "stars.")
```
